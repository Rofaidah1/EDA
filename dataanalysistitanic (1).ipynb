```python
import requests
import pandas as pd

# function تجيب التفاصيل من API
def get_car_details(vin):
    url = f"https://vpic.nhtsa.dot.gov/api/vehicles/DecodeVin/{vin}?format=json"
    response = requests.get(url)
    if response.status_code == 200:
        results = response.json()["Results"]
        return {
            "make": next((r["Value"] for r in results if r["Variable"] == "Make"), None),
            "model": next((r["Value"] for r in results if r["Variable"] == "Model"), None),
            "trim": next((r["Value"] for r in results if r["Variable"] == "Trim"), None),
            "body": next((r["Value"] for r in results if r["Variable"] == "Body Class"), None),
            "transmission": next((r["Value"] for r in results if "Transmission" in r["Variable"]), None),
        }
    return None


# نحدث الصفوف اللي فيها NaN بس
for idx, row in data.iterrows():
    if pd.isna(row['model']) or pd.isna(row['trim']) or pd.isna(row['body']) or pd.isna(row['transmission']):
        details = get_car_details(row['vin'])
        if details:
            if pd.isna(row['model']):
                data.at[idx, 'model'] = details['model']
            if pd.isna(row['trim']):
                data.at[idx, 'trim'] = details['trim']
            if pd.isna(row['body']):
                data.at[idx, 'body'] = details['body']
            if pd.isna(row['transmission']):
                data.at[idx, 'transmission'] = details['transmission']
```
